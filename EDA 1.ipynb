```python
# to convert the size column to numerical ---> replace M (mega byte) with 000 (1Mb = 1000 Kb), and remove k(kilo byte)
```

```python
def size_update(n):
    if n[-1] == 'M':
        return n.replace('M', '000')
    else:
        return n.replace('k', '')
```

```python
df['Size'] = df['Size'].apply(size_update)
```
